```python
import requests

def import_data_files():
  r = requests.get('https://raw.githubusercontent.com/anyoneai/notebooks/main/customers_and_orders/data/customers.csv')
  with open('./sample_data/customers.csv', 'wb') as f:
    f.write(r.content)

  r = requests.get('https://raw.githubusercontent.com/anyoneai/notebooks/main/customers_and_orders/data/orders.csv')
  with open('./sample_data/orders.csv', 'wb') as f:
    f.write(r.content)

import_data_files()
print("Customers and orders CSV files have been added './sample_data'")
```

```python
import csv

datafile = "./sample_data/customers.csv"
customers_list = []

with open(datafile, 'r') as fl:
    # Using DictReader
    csvreader = csv.DictReader(fl)

    for row in csvreader:
        # We use .get() or check the exact casing from your debug print
        # TRAP: Some CSVs use 'CustomerId', others 'CustomerID', others 'customer_id'
        clean_row = {
            'CustomerID': row.get('CustomerId') or row.get('CustomerID'),
            'FirstName': (row.get('First Name') or row.get('FirstName', '')).strip(),
            'LastName': (row.get('Last Name') or row.get('LastName', '')).strip().title(),
            'City': row.get('City', '').strip().title(),
            'State': row.get('State', '').strip().upper()
        }
        customers_list.append(clean_row)

print(f"Successfully processed {len(customers_list)} customers.")
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
```

```python
from collections import Counter

# Creamos el contador de estados (solo para filas con estado válido)
state_counts = Counter(c['State'] for c in customers_list if c['State'] and c['State'].strip() != "")

# Q3: State with most customers
most_common_state = state_counts.most_common(1)[0] # Retorna (Estado, Cantidad)

print(f"State with most customers: {most_common_state[0]} ({most_common_state[1]} customers)")
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file
```

```python
from collections import Counter

# Creamos el contador de estados (solo para filas con estado válido)
state_counts = Counter(c['State'] for c in customers_list if c['State'] and c['State'].strip() != "")

# Q4: State with least customers
# .most_common() ordena de mayor a menor, así que el último [-1] es el menor
least_common_state = state_counts.most_common()[-1]
print(f"State with least customers: {least_common_state[0]} ({least_common_state[1]} customers)")
```

**Question 5:** What is the most common last name?

```python
# open file and init vars
# evaluate the contents of the file
```

```python
last_name_counts = Counter(c['LastName'] for c in customers_list if c['LastName'])

most_common_last_name = last_name_counts.most_common(1)[0]

print(f"Most common last name: {most_common_last_name[0]} ({most_common_last_name[1]} occurrences)")
```
